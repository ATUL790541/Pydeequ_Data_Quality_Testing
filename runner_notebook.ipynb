### Pydeequ runner

```python



import os
import sys
import shutil
import subprocess
 
import pytest
 
TEST_ROOT = "/Workspace/Users/atul.gupta2@asper.ai/dataplatform_automation_test/etl_test/"
RESULTS = "/tmp/allure-results"
REPORT_DIR = "/tmp/allure-report"
ALLURE_BIN = "/tmp/allure/bin/allure"
 
# nothing may be written under TEST_ROOT from this cluster
sys.dont_write_bytecode = True
sys.path.insert(0, TEST_ROOT)
 
shutil.rmtree(RESULTS, ignore_errors=True)
 
print("Starting ETL tests...")
exit_code = pytest.main([
    os.path.join(TEST_ROOT, "tests/test_pydeequ_implementation.py"),
    "-sv",
    "-p", "no:cacheprovider",
    f"--alluredir={RESULTS}",
])
print(f"pytest exit code: {exit_code}")
 
if not os.path.isdir(RESULTS) or not os.listdir(RESULTS):
    raise RuntimeError(f"no allure results in {RESULTS} (pytest exited {exit_code})")
 
result = subprocess.run(
    [ALLURE_BIN, "generate", RESULTS, "-o", REPORT_DIR, "--clean", "--single-file"],
    capture_output=True,
    text=True,
)
print(result.stdout)
print(result.stderr)
result.check_returncode()
 
report = os.path.join(REPORT_DIR, "index.html")
dest = f"{TEST_ROOT}/Data_Validation_Report.html"
try:
    shutil.copy(report, dest)
except OSError as e:
    dest = "/Volumes/<catalog>/<schema>/<volume>/Data_Validation_Report.html"
    print(f"workspace write failed ({e.__class__.__name__}), falling back to volume")
    shutil.copy(report, dest)
 
print(f"Report saved at: {dest}")
```
